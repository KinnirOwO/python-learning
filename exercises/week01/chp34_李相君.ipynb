### 第3章　练习 1

**题目：** 编写一个 Python 程序，对输入字符串执行以下两个操作：1. 通过字母 “P” 更新第 5 个索引来更新字符串；2. 在字符串末尾连接单词 “world”。

**答案/解析：**

- 字符串是**不可变**对象，不能写 `s[5] = "P"`，否则会报 `TypeError`。
- 正确做法是“拆开再拼接”：`s[:5] + "P" + s[6:]`，即索引 5 之前的部分 + 新字母 + 索引 5 之后的部分。
- 末尾连接用 `+`。
- 运行时输入 `Hello World`。

```python
s = input("请输入一个字符串：")

# 1. 字符串不可变，只能拼出一个新字符串来“更新”第 5 个索引
new_s = s[:5] + "P" + s[6:]
print("更新第5个索引后：", new_s)

# 2. 在末尾连接 "world"
result = new_s + "world"
print("末尾连接 world 后：", result)
```

**程序执行流程分析：**

1. `input()` 读入 `Hello World`，`s` 指向这个字符串。
2. `s[:5]` 取索引 0～4，得到 `"Hello"`；`s[6:]` 从索引 6 取到末尾，得到 `"World"`，正好跳过了原来索引 5 的空格。
3. 三段用 `+` 拼接成新字符串 `"HelloPWorld"`，赋值给 `new_s`。原来的 `s` 仍然是 `"Hello World"`，没有被修改，这体现了字符串的不可变性。
4. `new_s + "world"` 又生成一个新字符串 `"HelloPWorldworld"`，赋值给 `result` 并输出。

### 第3章　练习 2

**题目：** 编写一个 Python 程序，对输入字符串执行以下操作：1. 打印字符串的长度；2. 将字符串转换为大写字母并打印；3. 打印字符串中的最大字母字符。

**答案/解析：**

- 长度用内置函数 `len(s)`，空格也算一个字符。
- 转大写用字符串方法 `s.upper()`，它返回一个新字符串。
- 最大字符用 `max(s)`，按字符的编码大小比较：空格 < 数字 < 大写字母 < 小写字母。
- 运行时输入 `Hello World`。

```python
s = input("请输入一个字符串：")

print("字符串长度：", len(s))
print("转成大写：", s.upper())
print("最大的字符：", max(s))
```

**程序执行流程分析：**

1. `input()` 读入 `Hello World`，`s` 指向这个字符串。
2. `len(s)` 统计字符个数：10 个字母 + 1 个空格 = 11。
3. `s.upper()` 生成一个全部大写的新字符串 `"HELLO WORLD"` 并输出；`s` 本身仍是原样，因为字符串不可变。
4. `max(s)` 逐个比较字符的编码值：空格是 32，大写 `H`、`W` 分别是 72、87，小写 `d e l o r` 分别是 100、101、108、111、114。最大的是小写 `r`（114），所以输出 `r`。大写 `W` 虽然在字母表里更靠后，但大写字母的编码整体比小写字母小。

### 第3章　练习 3

**题目：** 编写一个 Python 程序，对输入字符串执行以下操作：1. 使用字符串函数对 3 行段落进行对齐；2. 哪个字符串函数用于“在子字符串的最后一次出现时对字符串进行分区，并返回一个三元组，其中部分在分隔符之前，部分在分隔符之后”；3. 使用字符串函数计算下面给定段落中的行数。

**答案/解析：**

1. 对齐使用 `ljust()`（左对齐）、`center()`（居中）、`rjust()`（右对齐），格式为 `字符串.方法(宽度, 填充字符)`。
2. 答案是 **`rpartition()`**。它从右边开始查找分隔符，在**最后一次出现**的位置切开，返回 `(分隔符之前, 分隔符, 分隔符之后)` 三元组。与之对应的 `partition()` 是在**第一次出现**的位置切开。
3. 行数 = 换行符 `\n` 的个数 + 1，用 `count("\n") + 1` 计算；也可以用 `splitlines()` 拆成列表后用 `len()` 统计。题目中的段落按课本排版分为 4 行，用三重引号原样保存。

```python
# 1. 对 3 行文字进行对齐（宽度 20，用 * 填充便于观察）
line1 = "Python"
line2 = "is"
line3 = "fun"
print("--- 左对齐 ljust ---")
print(line1.ljust(20, "*"))
print(line2.ljust(20, "*"))
print(line3.ljust(20, "*"))
print("--- 居中 center ---")
print(line1.center(20, "*"))
print(line2.center(20, "*"))
print(line3.center(20, "*"))
print("--- 右对齐 rjust ---")
print(line1.rjust(20, "*"))
print(line2.rjust(20, "*"))
print(line3.rjust(20, "*"))
```

```python
# 2. 在分隔符最后一次出现的位置分成三段：rpartition()
text = "a-b-c-d"
print("rpartition：", text.rpartition("-"))
print("partition 对比：", text.partition("-"))
```

```python
# 3. 计算段落的行数
paragraph = """条件编程或分支是你每天、每时每刻都在做的事情。这是关于评估条件的，如果绿灯亮，我就可
以过马路，如果下雨，我就带上雨伞，如果上班迟到，我就给经理打电话。主要工具是 if 语
句，它有不同的形式和颜色，但它基本上是计算一个表达式，并根据结果选择要执行的代码部
分。像往常一样，让我们看一个例子。"""
print("换行符个数：", paragraph.count("\n"))
print("行数（count 换行符 + 1）：", paragraph.count("\n") + 1)
print("行数（splitlines）：", len(paragraph.splitlines()))
```

**程序执行流程分析：**

1. **对齐：** `"Python".ljust(20, "*")` 先把字符串放在宽度为 20 的格子最左边，剩下 20 − 6 = 14 个位置用 `*` 补在右边；`rjust` 补在左边；`center` 把补的位置分到两边。三个方法都返回新字符串，原字符串不变。
2. **rpartition：** 程序从 `"a-b-c-d"` 的右边开始找 `"-"`，最后一个 `-` 在 c 和 d 之间，于是切成三元组 `('a-b-c', '-', 'd')`。`partition` 从左边找到第一个 `-`，切成 `('a', '-', 'b-c-d')`，两者正好相反。
3. **行数：** 三重引号让字符串内部保留了 3 个换行符。`count("\n")` 返回 3，加 1 得到 4 行；`splitlines()` 按换行把段落拆成 4 个元素的列表，`len()` 得到 4，两种方法结果一致。

### 第3章　附加练习

**题目：** 编写一个 Python 程序，对输入字符串执行以下操作：1. 将字符串重复 3 次并打印；2. 使用 `format()` 函数将字符串从第二个索引截断为第六个索引。

**答案/解析：**

- 重复用 `*` 运算符：`s * 3`。
- 在 `format()` 中，`{:.n}` 用于字符串时表示“**最多保留 n 个字符**”。先用 `s[2:]` 从索引 2 开始取，再用 `{:.4}` 保留 4 个字符，就得到索引 2、3、4、5，与切片 `[2:6]` 含头不含尾的规则一致。
- 如果要求包括索引 6，则改用 `{:.5}`，代码中两种都给出。
- 运行时输入 `Programming`。

```python
s = input("请输入一个字符串：")

# 1. 重复 3 次
print("重复3次：", s * 3)

# 2. 用 format() 截取索引 2 到 6
print("索引2到6（不含6）：", "{:.4}".format(s[2:]))
print("索引2到6（含6）：", "{:.5}".format(s[2:]))
```

**程序执行流程分析：**

1. `input()` 读入 `Programming`，各字符的索引为：P0 r1 o2 g3 r4 a5 m6 m7 i8 n9 g10。
2. `s * 3` 生成一个把原字符串重复 3 遍的新字符串并输出。
3. `s[2:]` 先得到从索引 2 开始的子串 `"ogramming"`。
4. `"{:.4}".format(...)` 把这个子串填进 `{}`，`.4` 规定最多保留 4 个字符，所以输出 `ogra`（索引 2～5）；`{:.5}` 保留 5 个字符，输出 `ogram`（索引 2～6）。

### 第4章　练习 1

**题目：** 编写一个 Python 程序来执行以下列表中的操作。列表 = `['John', 'Allen', 'Black', 'Raj', 'Krish', 'Peter', 'Mike']`。1. 在列表末尾添加“安”一词；2. 在第三个索引处插入 “Ricky” 一词；3. 打印列表中单词 “Peter” 的索引。

**答案/解析：**

- 末尾添加用 `append("安")`。
- 在索引 3 插入用 `insert(3, "Ricky")`，原索引 3 及之后的元素都会向后移一位。
- 查找索引用 `index("Peter")`，返回第一次出现的位置。
- 三步按顺序执行后，`Peter` 的索引从 5 变为 **6**。

```python
names = ['John', 'Allen', 'Black', 'Raj', 'Krish', 'Peter', 'Mike']
print("原列表：", names)
print("Peter 原来的索引：", names.index("Peter"))

# 1. 在列表末尾添加 "安"
names.append("安")
print("append 后：", names)

# 2. 在索引 3 处插入 "Ricky"
names.insert(3, "Ricky")
print("insert 后：", names)

# 3. 打印 "Peter" 的索引
print("Peter 现在的索引：", names.index("Peter"))
```

**程序执行流程分析：**

1. 创建列表 `names`，共 7 个元素，`Peter` 在索引 5。
2. `append("安")` 直接**修改原列表**，在末尾加入一个元素，列表变为 8 个元素。列表是可变对象，所以不需要重新赋值。
3. `insert(3, "Ricky")` 把 `Ricky` 放到索引 3，原来索引 3 的 `Raj` 以及后面的 `Krish`、`Peter`、`Mike`、`安` 依次向后移动一位，列表变为 9 个元素。
4. 因为 `Peter` 被向后挤了一位，`index("Peter")` 返回 6。这说明操作顺序会影响结果：如果先查索引再插入，得到的是 5。

### 第4章　练习 2

**题目：** 编写一个 Python 程序，在下面给出的元组中执行以下操作。`coral = ('blue coral', 'staghorn coral', 'pillar coral', 'elkhorn coral')`。1. 打印第三个索引值；2. 按索引编号范围打印元组切片 `[1:3]`；3. 打印元组的长度。

**答案/解析：**

- “第三个索引”是 `coral[3]`，即 `elkhorn coral`。（注意它不同于“第三个元素” `coral[2]`。）
- 切片 `[1:3]` 取索引 1 和 2，结果仍然是元组。
- 长度用 `len(coral)`，为 4。

```python
coral = ('blue coral', 'staghorn coral', 'pillar coral', 'elkhorn coral')

print("coral[3] =", coral[3])
print("coral[1:3] =", coral[1:3])
print("切片结果的类型：", type(coral[1:3]))
print("元组长度 =", len(coral))
```

**程序执行流程分析：**

1. 创建元组 `coral`，4 个元素的索引分别是 0、1、2、3。
2. `coral[3]` 取索引 3，得到 `elkhorn coral`。
3. `coral[1:3]` 取索引 1、2（不含 3），得到 `('staghorn coral', 'pillar coral')`。`type()` 显示结果是 `tuple`，说明元组切片得到的还是元组。
4. `len(coral)` 返回元素个数 4。整个过程只“读取”元组、没有修改它，符合元组不可变的特点。

### 第4章　附加练习

**题目：** 编写一个 Python 程序，在下面给出的字典中执行以下操作。`sammy = {'username': 'sammy-shark', 'online': True, 'followers': 987}`。1. 向字典中添加一个键，然后打印新字典；2. 删除 “online” 键对应的键值对，并打印新字典。

**答案/解析：**

- 添加键使用 `字典[新键] = 值`：键不存在时就是新增。这里添加 `'location': 'Haikou'`。
- 删除键值对使用 `del 字典[键]`；也可以用 `pop(键)`，区别是 `pop` 会返回被删除的值。

```python
sammy = {'username': 'sammy-shark', 'online': True, 'followers': 987}
print("原字典：", sammy)

# 1. 添加一个新键
sammy['location'] = 'Haikou'
print("添加键后：", sammy)

# 2. 删除 "online" 键值对
del sammy['online']
print("删除 online 后：", sammy)
```

**程序执行流程分析：**

1. 创建字典 `sammy`，包含 3 个键值对。
2. 执行 `sammy['location'] = 'Haikou'` 时，Python 先查找键 `'location'`，发现不存在，于是新增这个键值对，字典变为 4 项。
3. `del sammy['online']` 找到键 `'online'`，把它和对应的值 `True` 一起删除，字典变为 3 项。
4. 字典是可变对象，两次操作都直接修改了原字典，所以打印 `sammy` 就能看到变化。
